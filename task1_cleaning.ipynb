Cell 1: Import and view the data 

```python
import pandas as pd

df = pd.read_csv("data/Dataset.csv")
print(df.shape)
df.head()
```

Cell 2: Check structure

```python
df.info()
```

Cell 3: Find missing values


```python
print(df.isnull().sum())
print((df == "Not Given").sum())
```

Cell 4: Handle missing values

```python
df["director"] = df["director"].replace("Not Given", "Unknown")
df["country"] = df["country"].replace("Not Given", "Unknown")
print((df == "Not Given").sum().sum())
```

Cell 5: Check duplicates

```python
print("Exact duplicates:", df.duplicated().sum())
dups = df[df.duplicated(subset=["title", "type", "release_year"], keep=False)]
dups
```

Cell 6: Remove duplicates

```python
before = len(df)
df = df.drop_duplicates(subset=["title", "type", "release_year"], keep="first")
print("Rows removed:", before - len(df))
```

Cell 7: Strip extra spaces

```python
for col in df.columns:
    if col != "release_year":
        df[col] = df[col].str.strip()
```

Cell 8: Standardize columns

```python
df["rating"] = df["rating"].str.upper()
df["date_added"] = pd.to_datetime(df["date_added"], format="%m/%d/%Y")
df["year_added"] = df["date_added"].dt.year
df["month_added"] = df["date_added"].dt.month_name()

df["duration_value"] = df["duration"].str.extract(r"(\d+)").astype(int)
df["duration_unit"] = df["duration"].str.extract(r"([A-Za-z]+)$")
```

Cell 9: Verify the standardization

```python
print(df["type"].unique())
print(df["rating"].unique())
print(df["country"].nunique(), "unique countries")
df.dtypes
```

Cell 10: Export

```python
df.to_csv("data/netflix_cleaned.csv", index=False)
print(df.shape)
df.head()
```

Cell 11: Final quality check

```python
print("Total missing:", df.isnull().sum().sum())
print("Unknown directors:", (df["director"] == "Unknown").sum())
print("Unknown countries:", (df["country"] == "Unknown").sum())
df["duration_unit"].value_counts()
```
